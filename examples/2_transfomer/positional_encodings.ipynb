```python
import numpy as np
```

```python
# Example sentence (tokens can just be integers for simplicity)
sentence = ["ChatGPT", "is", "great", "at", "explaining", "things", "clearly"]
vocab_size = 10000    # just for demonstration
embedding_dim = 16    # small dimension for illustration
```

```python
# 1️⃣ Create random token embeddings (simulating a real embedding layer)
np.random.seed(0)
token_embeddings = np.random.rand(len(sentence), embedding_dim)
print("Token embeddings:\n", token_embeddings, "\n")
```

```python
# 2️⃣ Positional encoding function (sinusoidal, as in Vaswani et al.)
def positional_encoding(seq_len, d_model):
    PE = np.zeros((seq_len, d_model))
    position = np.arange(seq_len)[:, np.newaxis]
    div_term = np.exp(np.arange(0, d_model, 2) * -(np.log(10000.0) / d_model))
    PE[:, 0::2] = np.sin(position * div_term)
    PE[:, 1::2] = np.cos(position * div_term)
    return PE
```

```python
# 3️⃣ Compute positional encodings
pos_encoding = positional_encoding(len(sentence), embedding_dim)
print("Positional encodings:\n", pos_encoding, "\n")
```

```python
# 4️⃣ Add positional encoding to token embeddings
emb_with_pos = token_embeddings + pos_encoding
print("Embeddings + positional encoding:\n", emb_with_pos)
```
